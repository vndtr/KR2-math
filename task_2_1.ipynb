## Алгебраическая, тригонометрическая и показательная формы

### Условие:
Представьте в алгебраической, тригонометрической и показательной формах выражения:

$$
a)\ \frac{(5+i)(7-6i)}{3+i};\qquad b)\ \frac{(1+i)^5}{(1-i)^3};\qquad c)\ \frac{(-1+i\sqrt3)^{15}}{(1-i)^{20}}+\frac{(-1-i\sqrt3)^{15}}{(1+i)^{20}}.
$$

### Теория:
Для $z=x+iy\ne0$ положим $r=|z|=\sqrt{x^2+y^2}$ и $\varphi=\operatorname{Arg}z$. Тогда
- алгебраическая форма: $z=x+iy$;
- тригонометрическая форма: $z=r(\cos\varphi+i\sin\varphi)$;
- показательная форма: $z=re^{i\varphi}$ (формула Эйлера).

Степень вычисляется по формуле Муавра: $z^n=r^n(\cos n\varphi+i\sin n\varphi)$. Частное: модули делятся, аргументы вычитаются.

### Решение вручную

**a)** Сначала сократим дробь $\dfrac{5+i}{3+i}$ (домножим на сопряжённое к знаменателю):
$$
\frac{5+i}{3+i}=\frac{(5+i)(3-i)}{9+1}=\frac{15-5i+3i+1}{10}=\frac{8-i}{5}.
$$
Теперь умножаем на $7-6i$:
$$
\frac{(8-i)(7-6i)}{5}=\frac{56-48i-7i+6i^2}{5}=\frac{50-55i}{5}=10-11i.
$$
Модуль $r=\sqrt{10^2+11^2}=\sqrt{221}$. Точка лежит в IV четверти, поэтому $\varphi=-\operatorname{arctg}\dfrac{11}{10}$.

**b)** Используем $(1+i)^2=2i$ и $(1-i)^2=-2i$:
$$
(1+i)^5=(2i)^2(1+i)=-4(1+i),\qquad (1-i)^3=(-2i)(1-i)=-2-2i=-2(1+i).
$$
Следовательно, $\dfrac{(1+i)^5}{(1-i)^3}=\dfrac{-4(1+i)}{-2(1+i)}=2$. Здесь $r=2$, $\varphi=0$.

**c)** Запишем числа в показательной форме:
$-1\pm i\sqrt3=2e^{\pm 2\pi i/3}$, а $(1\pm i)^2=\pm2i$, откуда $(1-i)^{20}=(-2i)^{10}=2^{10}i^{10}=-1024$ и $(1+i)^{20}=(2i)^{10}=-1024$.

По формуле Муавра $(-1\pm i\sqrt3)^{15}=2^{15}e^{\pm10\pi i}=2^{15}=32768$. Значит, каждое слагаемое равно $\dfrac{32768}{-1024}=-32$, а сумма равна $-64$.
Модуль $64$, аргумент $\pi$.

### Проверка в `sympy` (точные вычисления)

```python
from sympy import I, sqrt, simplify, expand, Abs, arg, N, atan, Rational

def show(name, expr):
    z = simplify(expand(expr))
    r, phi = simplify(Abs(z)), simplify(arg(z))
    print(f'{name}  z = {z}')
    print(f'    r = {r} ≈ {N(r, 8)},   φ = {phi} ≈ {N(phi, 8)}')
    print(f'    тригонометрическая: {r}·(cos({phi}) + i·sin({phi}))')
    print(f'    показательная:      {r}·exp(i·({phi}))')

show('a)', (5 + I)*(7 - 6*I)/(3 + I))
show('b)', (1 + I)**5/(1 - I)**3)
show('c)', (-1 + I*sqrt(3))**15/(1 - I)**20 + (-1 - I*sqrt(3))**15/(1 + I)**20)
print()
print('Проверка φ для a):', N(-atan(Rational(11, 10)), 8))
```

### Численная проверка: перевод из $(r,\varphi)$ обратно в алгебраическую форму

```python
import cmath

s3 = 3**0.5
values = {
    'a': (5 + 1j)*(7 - 6j)/(3 + 1j),
    'b': (1 + 1j)**5/(1 - 1j)**3,
    'c': (-1 + 1j*s3)**15/(1 - 1j)**20 + (-1 - 1j*s3)**15/(1 + 1j)**20,
}
for k, v in values.items():
    r, phi = cmath.polar(v)
    trig = complex(r*cmath.cos(phi), r*cmath.sin(phi))
    expo = r*cmath.exp(1j*phi)
    print(f'{k}) z = {v.real:.6f}{v.imag:+.6f}i | r = {r:.6f} | φ = {phi:.6f} | '
          f'trig = {trig.real:.6f}{trig.imag:+.6f}i | exp = {expo.real:.6f}{expo.imag:+.6f}i')
```

### Ответ:

| | алгебраическая | тригонометрическая | показательная |
|---|---|---|---|
| **a)** | $10-11i$ | $\sqrt{221}\,(\cos\varphi+i\sin\varphi)$, $\varphi=-\operatorname{arctg}\frac{11}{10}\approx-0{,}8330$ | $\sqrt{221}\,e^{-i\operatorname{arctg}(11/10)}$ |
| **b)** | $2$ | $2(\cos0+i\sin0)$ | $2e^{i\cdot0}$ |
| **c)** | $-64$ | $64(\cos\pi+i\sin\pi)$ | $64e^{i\pi}$ |
