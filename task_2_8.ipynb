# Задание 2.8
## Определитель $A+\lambda I$ для корня 14-й степени из единицы

### Условие:
Пусть $\lambda$ — комплексный корень степени $14$ из $1$ с положительной мнимой частью и максимальной среди таких корней действительной частью. Даны
$$
I=\begin{pmatrix}1&0\\0&1\end{pmatrix},\qquad A=\begin{pmatrix}7&5\\6&2\end{pmatrix}.
$$
Найдите $\det(A+\lambda I)$ и укажите: 1) его действительную часть; 2) его мнимую часть.

### Решение:
**Выбор $\lambda$.** Корни: $\lambda_k=\cos\dfrac{2\pi k}{14}+i\sin\dfrac{2\pi k}{14}=e^{i\pi k/7}$, $k=0,\dots,13$. Положительную мнимую часть имеют $k=1,\dots,6$ (углы лежат в интервале $(0,\pi)$). На $(0,\pi)$ косинус убывает, значит, наибольшая действительная часть у наименьшего угла:
$$
\lambda=e^{i\pi/7}.
$$
**Определитель.** Матрица $A+\lambda I=\begin{pmatrix}7+\lambda&5\\6&2+\lambda\end{pmatrix}$, поэтому
$$
\det(A+\lambda I)=(7+\lambda)(2+\lambda)-5\cdot6=\lambda^2+9\lambda-16.
$$
С учётом $\lambda^2=e^{2\pi i/7}$ получаем
$$
\operatorname{Re}=\cos\frac{2\pi}7+9\cos\frac\pi7-16,\qquad
\operatorname{Im}=\sin\frac{2\pi}7+9\sin\frac\pi7.
$$

```python
import numpy as np

n = 14
roots = np.exp(2j*np.pi*np.arange(n)/n)
upper = roots[roots.imag > 1e-12]
print('корни с Im > 0:')
for z in upper:
    print(f'   {z.real: .6f} {z.imag:+.6f}i')

lam = upper[np.argmax(upper.real)]
print('\nвыбранный λ =', lam, '   (exp(iπ/7) =', np.exp(1j*np.pi/7), ')')
```

```python
A = np.array([[7, 5], [6, 2]], dtype=complex)
d_numpy = np.linalg.det(A + lam*np.eye(2))
d_formula = lam**2 + 9*lam - 16
d_trig = complex(np.cos(2*np.pi/7) + 9*np.cos(np.pi/7) - 16,
                 np.sin(2*np.pi/7) + 9*np.sin(np.pi/7))

print('numpy.linalg.det    :', d_numpy)
print('λ² + 9λ − 16        :', d_formula)
print('через cos и sin     :', d_trig)
```

### Точная проверка (`sympy`)

```python
from sympy import Matrix, exp, I, pi, eye, N, expand_complex, re, im

lam_s = exp(I*pi/7)
d = (Matrix([[7, 5], [6, 2]]) + lam_s*eye(2)).det()
d = expand_complex(d.expand())
print('Re det =', N(re(d), 12))
print('Im det =', N(im(d), 12))
```

### Ответ:

$\lambda=e^{i\pi/7}$, $\ \det(A+\lambda I)=\lambda^2+9\lambda-16\approx-7{,}26779+4{,}68679\,i$.

1) действительная часть: $\cos\dfrac{2\pi}7+9\cos\dfrac\pi7-16\approx\mathbf{-7{,}26779}$;

2) мнимая часть: $\sin\dfrac{2\pi}7+9\sin\dfrac\pi7\approx\mathbf{4{,}68679}$.
